```python
import pandas as ps
df=ps.read_csv('.\BreastCancer.csv')
df.head()
```

```python
x=df.drop('outcome',axis=1)
y=df['outcome']
x
```

```python
import numpy as np
x=np.array(x)
y=np.array(y)
x
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x=scaler.fit_transform(x)
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
x_train.shape,y_train.shape
```

## 1.Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB
gnb=GaussianNB()
gnb.fit(x_train,y_train)
```

```python
y_pred_test=gnb.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,recall_score,precision_score
acc_gnb=accuracy_score(y_true=y_test,y_pred=y_pred_test)
r_gnb=recall_score(y_true=y_test,y_pred=y_pred_test)
p_gnb=precision_score(y_true=y_test,y_pred=y_pred_test)
print('acc_gnb:',acc_gnb)
print('r_gnb',r_gnb)
print('p_gnb',p_gnb)
```
